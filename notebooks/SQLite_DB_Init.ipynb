```julia
# Good video to start from (some syntax is deprecated): https://www.youtube.com/watch?v=qUrtLJcehE0
# DDL (define) - Create, Alter, Drop          - Command: SQLite.DB()
# DML (manipulate) - Insert, Update, Delete   - Command: SQLite.execute(
# DQL (query) - Select                        - Command: SQLite.Query()


using SQLite
# Dataframes adds ability to visualize nicer
using DataFrames


# Creates or open a file named "test_database.sqlite" in the current directory
db = SQLite.DB("test_simple_grid_database.sqlite")
```

```julia
# Create all tables
# Primary key is created by default in SQLite, access it using ROWID

# Subscribers
SQLite.execute(db, "CREATE TABLE IF NOT EXISTS subscribers(
    account_id INTEGER PRIMARY KEY,
    name TEXT, 
    email TEXT)
    ")

# Buses
SQLite.execute(db, "CREATE TABLE IF NOT EXISTS buses(
    bus_key_id INTEGER PRIMARY KEY,
    bus_id INTEGER, 
    address TEXT,
    account_id INTEGER,
    FOREIGN KEY(account_id) REFERENCES subscribers(account_id))
    ")

# Records
SQLite.execute(db, "CREATE TABLE IF NOT EXISTS records(
    record_id INTEGER PRIMARY KEY,
    record_time DATETIME, 
    power_quality REAL,
    status INTEGER,
    bus_id INTEGER,
    FOREIGN KEY(bus_id) REFERENCES buses(bus_id))
    ")

# Global Records
SQLite.execute(db, "CREATE TABLE IF NOT EXISTS globalRecords(
    global_record_id INTEGER PRIMARY KEY,
    record_time DATETIME, 
    power_quality REAL,
    num_islands INTEGER,
    converges INTEGER)
    ")
```

```julia
SQLite.execute(db, "DROP TABLE subscribers")
```
